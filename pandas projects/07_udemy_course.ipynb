# 14. Display 10 Most Popular Courses As Per Number of Subscribers

Here using Hue might be confusing....
as hue is used to splits data into group.


since y and hue both has same column name, each course becomes its own unique group. As a result, Seaborn assigns a different color to every bar instead of grouping multiple bars together.


```python
popu_10=df.sort_values(by="num_subscribers",ascending= False).head(10)

sns.barplot(x="num_subscribers",
             y="course_title",
             hue="course_title",
             data= popu_10,
             palette= "coolwarm",
             legend=False)
```

# 15. Find The Course Which Is Having The Highest Number of Reviews.

```python
print(df.sort_values(by="num_reviews",ascending=False).head(1)['course_title'])
plt.figure(figsize=(10,4))
sns.barplot(x='subject',y='num_reviews',hue="subject",data= df,palette="rocket",legend= False)
plt.show()
```

# 18. Display 10 Most Popular Python Courses As Per Number of Subscribers


```python
top_10=df[df["course_title"].str.contains("python",
                                          case=False)].sort_values(by="num_subscribers",
                                                                   ascending=False).head(10)

sns.barplot(x="num_subscribers",
            y="course_title",
            hue="course_title",
            data=top_10,
            palette="rocket"
            )

plt.show()
```

# 19. In Which Year The Highest Number of Courses Were Posted?


```python
df.columns
```

Here first converting it into date time format as panda understands T and Z as time and zone

```python
df["published_timestamp"]=pd.to_datetime(df["published_timestamp"])
df["Year"] = df["published_timestamp"].dt.year

df["Year"]
```

```python
df["Year"].value_counts()

sns.countplot(x="Year",data=df,hue="Year",palette="coolwarm")
plt.legend(loc="upper left")
plt.show()
```

# 20. Display Category-Wise Count of Posted Subjects [Year Wise] 

unstack() says:

Take the innermost index level and turn it into columns.

fill_value=0 replaces NaN with 0.

```python
df.groupby(["Year","subject"]).size().unstack(fill_value=0)
```
